## Node Name
`detect_duckiebot_node.py`

---

## Purpose
Dieser Node implementiert die Spurverfolgung für den Duckiebot. Er verarbeitet Kamerabilder, erkennt die Spur und steuert den Duckiebot innerhalb der Fahrbahnmarkierungen.
## Internal Logic
1. Kamerabild wird abonniert und vorverarbeitet (z. B. Graustufen, Canny-Filter, etc.).
2. Lane-Detection-Algorithmus erkennt die Spur (z. B. Hough-Lines oder Deep Learning).
3. Mit Hilfe eines PID-Reglers werden Lenk- und Geschwindigkeitsbefehle berechnet.
4. Die berechneten Werte werden über `/duckiebot/car_cmd` veröffentlicht.

---

## Input Topics
| Topic Name                      | Type                         | Description                                           |
|---------------------------------|-----------------------------|-------------------------------------------------------|
| `/camera_node/image/raw`        | `sensor_msgs/Image`         | Rohbild der Frontkamera zur Verarbeitung der Spur.    |
| `/lane_pose`                    | `duckietown_msgs/LanePose`  | Position des Duckiebots innerhalb der Fahrspur.       |

---

## Output Topics
| Topic Name                      | Type                            | Description                                                |
|---------------------------------|---------------------------------|------------------------------------------------------------|
| `/duckiebot/car_cmd`            | `duckietown_msgs/Twist2DStamped` | Bewegungsbefehl für Geschwindigkeit (v) und Lenkung (omega).|

---

## Parameters
| Name    | Type   | Default Value | Description                                     |
|---------|--------|---------------|-------------------------------------------------|
| `k_p`   | float  | 1.0           | Proportionalfaktor für die Spurregelung.         |
| `k_d`   | float  | 0.1           | Dämpfungsfaktor für die Spurregelung.            |
